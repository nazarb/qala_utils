```python
from ultralytics import YOLO
import torch

model = YOLO('yolov9_qanat.yaml')


# Now train
train_results = model.train(
    data="./datasets/qanats_256_synt_G1_AFG1_pairs_single_4.yaml",
    epochs=300,
    imgsz=256,
    batch=32,
    augment=True,
    translate=0.0,
    scale=0.0,
    mosaic=0.0,
    device="0"
)
```
